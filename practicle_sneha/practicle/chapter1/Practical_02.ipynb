# Practical 2
### Aim
To demonstrate creation and manipulation of list, tuple, set, dictionary and string using their built-in methods.

### 1. List

```python
friends = ["Meet", "Harsh", "Taksh", "Aryan"]
print(friends)
print(friends[0])
print(friends[-1])

friends.append("Adi")
print(friends)
friends.insert(1, "Darsh")
print(friends)
friends.remove("Adi")
print(friends)
x = friends.pop()
print(x)
print(friends)
friends.sort()
print(friends)
friends.reverse()
print(friends)
print(len(friends))
print(friends.count("Darsh"))
print(friends.index("Meet"))
```

### 2. Tuple

```python
subjects = ("Python", "CN", "MPI", "PDS", "PM")
print(subjects)
print(subjects[0])
print(subjects[-1])
print(len(subjects))
print(subjects.count("Python"))
print(subjects.index("PDS"))
print(subjects[1:4])
```

### 3. Set

```python
numbers = {10, 20, 30, 40, 20}
print(numbers)
numbers.add(50)
print(numbers)
numbers.update([60, 70])
print(numbers)
numbers.remove(30)
print(numbers)
numbers.discard(100)
print(numbers)
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
```

### 4. Dictionary

```python
student = {"name": "sneha", "roll_no": 41, "age": 21, "course": "Computer Engineering"}
print(student)
print(student["name"])
print(student.get("age"))
student["city"] = "Ahmedabad"
print(student)
student["age"] = 22
print(student)
student.pop("city")
print(student)
print(student.keys())
print(student.values())
print(student.items())
student.update({"age": 21, "marks": 85})
print(student)
```

### 5. String

```python
text = "Python for Data Science"
print(text)
print(len(text))
print(text[0])
print(text[0:6])
print(text.upper())
print(text.lower())
print(text.title())
print(text.capitalize())
print(text.replace("Python", "Python Programming"))
print(text.find("Data"))
print(text.count("o"))
print(text.startswith("Python"))
print(text.endswith("Science"))
a = "  hello  "
print(a.strip())
print(a.lstrip())
print(a.rstrip())
words = text.split(" ")
print(words)
print("-".join(words))
```
